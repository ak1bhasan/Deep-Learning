**Part A Answer:**

Difference between Sigmoid & ReLU activition function is The Sigmoid activation function outputs values between 0 & 1, is useful for binary classification. On the otherhand, ReLU outputs values from 0 to infinity, meaning it outputs only non negative values. Also, Sigmoid is commonly used in output layer, where ReLU is mainly used in hidden layer.


Relu is preferred in hidden layers because it activates only positive neurons and it is faster in training & better performance.

One major problem with Sigmoid is the vanishing gradient problem. During backpropagation, the gradients can become very small. As a result, the weights and biases are updated very slowly, so the network learns slowly.


**Part B Answer:**

Key difference between Batch Gradient Descent, Stochastic Gradient Descent (SGD), and Mini-batch Gradient Descent is :

Batch Gradient Descent uses the entire dataset to calculate the gradient & update weights once.

SGD uses one training example at a time to calculate the gradient & update weights

And Mini-batch uses a small batch of training examples to calculate the gradient & update weights.

Batch Gradient Descent gives a stable, accurate updates. SGD gives fast updates and use less memory and Mini-Batch provides a good balance between speed and stablity.

For Fashion-MNIST (60,000 images), I recommend Mini-Batch because it processes a small number of images at a time.

**Part C Answer:**

1. We prefer CrossEntropyLoss because it is suitable for multi-class classification.
2. CrossEntropyLoss already handles Softmax internally. Thats why we should not put nn.Softmax.

## Question 4: Define the Model (15 Marks)

Create a class `FashionClassifier` inheriting from `nn.Module`.

Use `nn.Sequential` with this architecture:

| Layer | Description |
|-------|-------------|
| Flatten | Convert 28×28 → 784 |
| Linear | 784 → 256 |
| ReLU | Activation |
| Linear | 256 → 128 |
| ReLU | Activation |
| Linear | 128 → 64 |
| ReLU | Activation |
| Linear | 64 → 10 |

> **Note**: Do NOT apply Softmax at the output layer.

After defining the class:
1. Create an instance of `FashionClassifier`.
2. Print the model architecture.

```python
# YOUR CODE HERE

class FashionClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 10)
        )

    def forward(self , features):
      output = self.network(features)
      return output
```

```python
model = FashionClassifier()
print(model)
```

## Question 5: Train the Model (25 Marks)

### 5.1 Set Up Training (5 Marks)
- Loss function: `nn.CrossEntropyLoss`
- Optimizer: `optim.SGD` with `lr=0.01`
- Number of epochs: **15**

```python
# YOUR CODE HERE

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD( model.parameters(), lr = 0.01 )
epochs = 15
```

### 5.2 Training Loop (15 Marks)

Write a complete training loop for **15 epochs**.

For each epoch:
- Train on `train_loader` and compute training loss
- Evaluate on `val_loader` (with `torch.no_grad()`) and compute validation loss
- Print: `Epoch {epoch+1}/{epochs} | Train Loss: {:.4f} | Val Loss: {:.4f} `

Store: `train_losses`, `val_losses`, (lists, one value per epoch).

```python
# YOUR CODE HERE

train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []


for epoch in range( epochs ) :
  running_loss = 0.0
  correct = 0
  total = 0

  for images, labels in train_loader :
    optimizer.zero_grad()

    outputs = model( images )

    loss = criterion( outputs, labels )

    loss.backward()

    optimizer.step()

    running_loss += loss.item()

    predicted = torch.argmax( outputs, dim = 1 )


    correct += ( predicted == labels ).sum().item()
    total += labels.size(0)

  val_loss = 0.0
  val_correct = 0
  val_total = 0

  with torch.no_grad():
    for images, labels in val_loader:
        outputs = model( images )
        loss = criterion( outputs, labels )
        val_loss += loss.item()

        predicted = torch.argmax( outputs, dim = 1 )
        val_correct += (predicted == labels).sum().item()
        val_total += labels.size(0)



  train_losses.append( running_loss / len(train_loader) )

  val_losses.append( val_loss / len(val_loader) )


  train_accuracies.append( 100 * correct / total )
  val_accuracies.append( 100 * val_correct / val_total )

  print(f'Epoch {epoch+1}/{epochs} | Train Loss: {train_losses[-1]:.4f} | Val Loss: {val_losses[-1]:.4f}')
```

```python
# --- TEST CELL (DO NOT MODIFY) ---
assert len(train_losses) == 15, f"Expected 15 epochs, got {len(train_losses)}"
assert len(val_losses) == 15
assert train_losses[-1] < train_losses[0], "Training loss should decrease"
print("✅ Q5.2 Passed!")
print(f"   Final Train Acc: {train_accuracies[-1]:.2f}%, Final Val Acc: {val_accuracies[-1]:.2f}%")
```

### 5.3 Test Evaluation (5 Marks)

Evaluate the model on the **test set**. Print the test_accuracy (required_variable)

Also, collect all test predictions and true labels into lists called `all_preds` and `all_labels` (we'll use them for analysis later).

Print: `Test Accuracy: {:.2f}%`

```python
# YOUR CODE HERE

test_correct = 0
test_total = 0

all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:

        outputs = model( images )

        predicted = torch.argmax( outputs, dim = 1 )

        test_correct += ( predicted == labels ).sum().item()
        test_total += labels.size(0)



        all_preds.extend( predicted )
        all_labels.extend(labels  )

test_accuracy = 100 * ( test_correct / test_total )
print(f'Test Accuracy: {test_accuracy:.2f}%')
```

```python
# --- TEST CELL (DO NOT MODIFY) ---
assert test_accuracy > 40.0, f"Test accuracy should be > 40%, got {test_accuracy:.2f}%"
assert len(all_preds) == len(X_test), "Predictions should cover all test samples"
print(f"✅ Q5.3 Passed! Test Accuracy: {test_accuracy:.2f}%")
```
