```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'x1' : [ 0, 0, 1, 1 ],
    'x2' : [ 0, 1, 0, 1 ],
    'y'  : [ 0, 0, 0, 1 ]
}

AND = pd.DataFrame(data)

AND
```

```python
plt.figure(figsize = (8,6))

sns.scatterplot(
    data = AND,
    x = 'x1',
    y = 'x2',
    hue = 'y'
)

intercept = 1.5

sns.lineplot(x=[0, intercept], y=[intercept, 0])

plt.xlabel('Figure 1')
plt.ylabel('Figure 2')
plt.title(f"Feature Space and Decision Boundary for 2-Feature AND Dataset")
plt.show()
```

```python
data = {
    'x1' : [ 0, 0, 1, 1 ],
    'x2' : [ 0, 1, 0, 1 ],
    'y'  : [ 0, 1, 1, 0 ]
}

XOR = pd.DataFrame(data)

XOR
```

```python
plt.figure(figsize = (8,6))

sns.scatterplot(
    data = XOR,
    x = 'x1',
    y = 'x2',
    hue = 'y'
)

sns.lineplot(x=[0, 1], y=[0, 1])
sns.lineplot(x=[0, 1], y=[1, 0])

plt.xlabel('Figure 1')
plt.ylabel('Figure 2')
plt.title("XOR Dataset's Feature Space")
plt.show()
```
