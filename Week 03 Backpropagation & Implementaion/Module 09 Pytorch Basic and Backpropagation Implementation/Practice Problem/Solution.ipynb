## 5. Manual Forward Pass of a Two-Layer Neural Network

A simple two-layer neural network can be written as:

$$
Z_1 = XW_1 + b_1
$$

$$
A_1 = f(Z_1)
$$

$$
Z_2 = A_1W_2 + b_2
$$

$$
A_2 = f(Z_2)
$$

Where:
- $Z_1$ = output of the first layer before activation
- $A_1$ = output of the first layer after activation
- $Z_2$ = output of the second layer before activation
- $A_2$ = final output
- $f$ = activation function

```python
Z1 = torch.matmul(X, W) + b
print(Z1)
```

### Forward Pass Calculation

Since:

$$
XW = 1.1
$$

and:

$$
b = 0.1
$$

we get:

$$
Z_1 = XW + b
$$

$$
Z_1 = 1.1 + 0.1 = 1.2
$$

If we use **ReLU** as the activation function:

$$
A_1 = \max(0, Z_1)
$$

$$
A_1 = \max(0, 1.2) = 1.2
$$

Therefore, the first layer's output is:

$$
Z_1 = 1.2
$$

$$
A_1 = 1.2
$$

**Key idea:** Matrix multiplication works because the columns of `X` match the rows of `W`:

$$
(1 \times 2)(2 \times 1) = (1 \times 1)
$$

## 2. Two-Layer Neural Network

You are given a two-layer neural network with the following parameters:

- Input $X$ is a column vector of size $(2 \times 1)$.
- $W_1$ is of shape $(2 \times 2)$.
- $b_1$ is of shape $(2 \times 1)$.
- $W_2$ is of shape $(2 \times 1)$.
- $b_2$ is of shape $(1 \times 1)$.

### Tasks:

**a)** Write the mathematical formula for $Z_1$.

**b)** Apply ReLU activation to compute $A_1$.

**c)** Write the formula for the final output $\hat{y}$.

**d)** If all weights are initialized to $0.1$ and biases are zero, compute the final output manually for the input:

$$
X =
\begin{bmatrix}
1.0 \\
2.0
\end{bmatrix}
$$

## Solution

### Given

- Input $X$ is a column vector of shape $(2 \times 1)$
- $W_1$ has shape $(2 \times 2)$
- $b_1$ has shape $(2 \times 1)$
- $W_2$ has shape $(2 \times 1)$
- $b_2$ has shape $(1 \times 1)$

### a) Formula for $Z_1$

Since $X$ is a column vector:

$$
Z_1 = W_1^T X + b_1
$$

### b) Apply ReLU

$$
A_1 = ReLU(Z_1)
$$

where:

$$
ReLU(x) = \max(0,x)
$$

### c) Formula for the final output

$$
\hat{y} = W_2^T A_1 + b_2
$$

### d) Manual Calculation

Given:

$$
X =
\begin{bmatrix}
1.0 \\
2.0
\end{bmatrix}
$$

All weights are initialized to $0.1$ and biases are zero:

$$
W_1 =
\begin{bmatrix}
0.1 & 0.1 \\
0.1 & 0.1
\end{bmatrix}
,\quad
b_1 =
\begin{bmatrix}
0 \\
0
\end{bmatrix}
$$

$$
W_2 =
\begin{bmatrix}
0.1 \\
0.1
\end{bmatrix}
,\quad
b_2 =
\begin{bmatrix}
0
\end{bmatrix}
$$

First, calculate $Z_1$:

$$
Z_1 = W_1^T X + b_1
$$

$$
Z_1 =
\begin{bmatrix}
0.1 & 0.1 \\
0.1 & 0.1
\end{bmatrix}
\begin{bmatrix}
1 \\
2
\end{bmatrix}
=
\begin{bmatrix}
0.3 \\
0.3
\end{bmatrix}
$$

Apply ReLU:

$$
A_1 =
\begin{bmatrix}
0.3 \\
0.3
\end{bmatrix}
$$

Finally:

$$
\hat{y} = W_2^T A_1 + b_2
$$

$$
\hat{y} = (0.1)(0.3) + (0.1)(0.3)
$$

$$
\boxed{\hat{y} = 0.06}
$$

**Final output: $0.06$**
